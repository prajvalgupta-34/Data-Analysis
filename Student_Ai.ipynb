```python
import os
from pathlib import Path

import kagglehub
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Download the dataset
path = kagglehub.dataset_download(
    "harshadapatil31/student-performance-and-study-habits-dataset"
)
print("Dataset folder:", path)

# Find CSV files in the downloaded folder
csv_files = list(Path(path).rglob("*.csv"))
if not csv_files:
    raise FileNotFoundError(f"No CSV files found in: {path}")

print("CSV files found:", [file.name for file in csv_files])

# Load the first CSV file
csv_path = csv_files[0]
df = pd.read_csv(csv_path)

print("Loaded:", csv_path.name)
print("Working directory:", os.getcwd())
print("Dataset shape:", df.shape)

display(df.head())
df.info()
```

```python
df['gender'].unique()
map_gender={'Male':1,'Female':2}
```

```python
df['gender']=df['gender'].map(map_gender)
```

```python
df['parental_education'].unique()
map_parental_education={'Bachelors':2, 'High School':1, 'Masters':3, 'PhD':4}
df['parental_education']=df['parental_education'].map(map_parental_education)
#df['parental_education']=df['parental_education'].map(map_parental_education).fillna(-5)
df=df.dropna(axis=0)
```

```python
df['internet_access'].unique()
map_internet_access={'Yes':2,'No':1}
df['internet_access']=df['internet_access'].map(map_internet_access)
```

```python
df['part_time_job'].unique()
map_part_time_job={'Yes':2,'No':1}
df['part_time_job']=df['part_time_job'].map(map_part_time_job)
```

```python
df['final_grade'].unique()
map_final_grade={'A':5,'B':4,'C':3,'D':2,'F':1}
df['final_grade']=df['final_grade'].map(map_final_grade)
```

```python
df['extracurricular_activities'].unique()
df['extracurricular_activities']=df['extracurricular_activities'].map({'Yes':2,'No':1})
```

```python
feature_names=['gender', 'study_time_hours', 'attendance_percent',
       'sleep_hours', 'parental_education', 'internet_access',
       'extracurricular_activities', 'part_time_job', 'previous_grade',
       'final_exam_score',]
```

```python
X=df[feature_names]
```

```python
y=df['final_grade']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=0)
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
best_k=0
best_score=0
for k in range(1,40):
    KNN=KNeighborsClassifier(n_neighbors=k,p=1)
KNN.fit(X_train_scaled,y_train)
```

```python
accuracy=KNN.score(X_test_scaled,y_test)
if accuracy > best_score:
        best_score = accuracy
        best_k = k

print("-" * 30)
print(f"Peak Performance: K = {best_k} achieved {best_score:.4f} accuracy")
```
